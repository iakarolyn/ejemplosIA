```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Perceptron
from sklearn.metrics import classification_report, accuracy_score

data = pd.read_csv('SMSSpamCollection', sep='\t', header=None, names=["validacion", "message"])

data['validacion'] = data['validacion'].map({'spam': 1, 'ham': 0}) # no es spam =0 y es spam =1


X = data['message']
y = data['validacion']

#se utiliza para convertir un conjunto de documentos de texto
#en una matriz de valores TF-IDF (Term Frequency-Inverse Document Frequency).
#Esta representación es útil para tareas de procesamiento de lenguaje natural (NLP),
#como clasificación de texto, clustering o búsqueda.
vectorizer = TfidfVectorizer(stop_words='english')
X_tfidf = vectorizer.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_tfidf, y, test_size=0.3, random_state=42)

model = Perceptron()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print ("prueba git")

nuevo_correo = ["Join our exciting raffle! Win amazing prizes. Buy your tickets here:"]
nuevo_correo_tfidf = vectorizer.transform(nuevo_correo)
prediccion = model.predict(nuevo_correo_tfidf)

if prediccion == 1:
    print("El correo es spam.")
else:
    print("El correo no es spam.")

 
import joblib


joblib.dump(model, 'modeloPerceptron.pkl')

```
